```python
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
from scipy.interpolate import UnivariateSpline
import numpy as np

meson_fields = np.loadtxt("meson_fields_pb.txt")
xi = -meson_fields[:,2] - meson_fields[:,1] + 0.5*meson_fields[:,3]
r = meson_fields[:,0]

spline = UnivariateSpline(r, xi, s=1)
der_spline = spline.derivative(n=1)

# Define the differential equation for T(r)
def dTdr(r, T):
    dlnxidr = der_spline(r)/spline(r)
    dT_dr = T**2 + T*dlnxidr
    return dT_dr

t_eval = np.linspace(r[-1],r[0],500)
sol = solve_ivp(dTdr,t_span=(r[-1],r[0]),y0=[1e-10],t_eval=t_eval)

# Plot results for each initial condition
plt.figure(figsize=(10, 6))
plt.plot(sol.t, sol.y[0])
#plt.plot(r, xi, label=f"T(r) with T(0) = {T_init}")
```

```python
# Figure out how to compute cov from scipy.least_squares
# Defining a linear model with predefined std deviations on the slope and intercept
# Will generate 1000 samples from these models using Monte Carlo and then fit the data and compute the covariance
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as sp
from scipy.stats import norm

# define linear model with parameters m,b and argument x
def model(m,b,x):
    return m*x+b

# define the means of the model paramaters and std deviations
M = 5.0
B = 2.0
sM = 0.5
sB = 0.1

# define the arguments and result arr
obs_x = np.linspace(0,5,20)
xval_raw = []
yval_raw = []
errs = []
means = []

# use monte carlo sampling to generate observable data set
for j in range(20):
    res = []
    for i in range(5):
        x = obs_x[j]
        r_m = np.random.normal(loc=M, scale=sM)
        r_b = np.random.normal(loc=B, scale=sB)
        y = model(r_m,r_b,x)
        xval_raw.append(x)
        res.append(y)
        yval_raw.append(y)

    # compute errors and means for the generated data
    errs.append(np.std(res))
    means.append(np.mean(res))

# plot the data to visualize
fig, axs = plt.subplots(1, 2, figsize=(12, 6))
axs[0].scatter(xval_raw,yval_raw,s=1)
axs[0].errorbar(obs_x,means,yerr=errs,color='red')
axs[1].hist(res,density=True)
x = np.linspace(means[-1] - 4*errs[-1], means[-1] + 4*errs[-1], 1000)
pdf = norm.pdf(x, means[-1], errs[-1])
axs[1].plot(x, pdf)

# Define initial guess for parameters and rescaling
x0 = [0.5, 0.4]
conv = [10.0, 5.0]

# define residuals function
def residuals(params):
    residuals = []
    m = params[0]
    b = params[1]
    for i in range(len(obs_x)):
        y = model(m*conv[0],b*conv[1],obs_x[i])
        res = (y - means[i])/errs[i]
        residuals.append(res)

    return np.array(residuals)

# Fit using least squares (Levenberg-Marquardt)
result = sp.least_squares(residuals,x0=x0,method='lm',diff_step=1e-4)
print("LSQ means: ", np.array(result.x)*np.array(conv))
J = np.array(result.jac)
hess = np.matmul(np.transpose(J),J)
cov = np.linalg.inv(hess)
scale = np.diag(conv)
cov = np.matmul(np.matmul(scale,cov),scale)
invcov = np.linalg.inv(cov)
print(invcov)

# Use monte carlo to verify least squares results
def metropolis(post0, postp, x_0, x_p, acc_counts, index):
    # metroplis hastings step
    r = np.random.uniform(0,1)
    a = postp/post0
    if (a>1):
        a=1.0
    if (r <= a):
        post0 = postp
        for k in range(2):
            x_0[k] = x_p[k] # accept the proposed changes
        acc_counts[index]+=1   # count how many times the changes are accepted
    return post0

# adaptive MCMC method to reach acceptance rates
def adaptive_width(iter,n_check,arate,acc_counts,stds,agoal,index):
    if ((iter+1)%n_check == 0):
        arate[index] = acc_counts[index]/n_check
        acc_counts[index] = 0
        if (arate[index] < agoal):
            stds[index] = 0.9*stds[index]      # if acceptance rate is too low then decrease the range
        elif (arate[index] > agoal):
            stds[index] = 1.1*stds[index]      # if acceptance rate is too high then increase the range
            
def MCMC(n_burnin, n_runs, x_0, stds):
    np.random.seed(0)
    x_p = np.empty_like(x_0)
    acc_counts = [0,0]
    arate = [0,0]
    agoal = 0.5

    chi2 = 0.0
    for k in range(len(obs_x)):
        chi2 = chi2 + (means[k] - model(x_0[0]*conv[0],x_0[1]*conv[1],obs_x[k]))**2/errs[k]**2
    post0 = np.exp(-chi2/2.0)

    for i in range(n_burnin):
        # one sweep
        for j in range(2):
            # get new proposed parameters
            for k in range(2):
                x_p[k] = x_0[k] # copy old values
            x_p[j] = np.random.normal(x_0[j],stds[j])      # change one param

            # compute chi square
            chi2 = 0.0
            for k in range(len(obs_x)):
                chi2 = chi2 + (means[k] - model(x_p[0]*conv[0],x_p[1]*conv[1],obs_x[k]))**2/errs[k]**2
            postp = np.exp(-chi2/2.0)

            # metroplis hastings step
            post0 = metropolis(post0,postp,x_0,x_p,acc_counts,j)

            # rate monitoring to adjust the width of the sampling
            adaptive_width(i,50,arate,acc_counts,stds,agoal,j)
            #if ((i+1)%50 == 0):
                #print(arate,stds)

            # print MCMC sweep
            #print(x_0[0], x_0[1])
    ##############################################
    # end of burn in

    # start MCMC runs
    samples = []
    for i in range(n_runs):
        # one sweep
        for j in range(2):
            # get new proposed parameters
            for k in range(2):
                x_p[k] = x_0[k] # copy old values
            x_p[j] = np.random.normal(x_0[j],stds[j])      # change one param

            # compute chi square
            chi2 = 0.0
            for k in range(len(obs_x)):
                chi2 = chi2 + (means[k] - model(x_p[0]*conv[0],x_p[1]*conv[1],obs_x[k]))**2/errs[k]**2
            postp = np.exp(-chi2/2.0)

            # metroplis hastings step
            post0 = metropolis(post0,postp,x_0,x_p,acc_counts,j)

            # print MCMC sweep and store samples
        samples.append(x_0.copy())
    
    return samples


samp = np.array(MCMC(10000,50000,x0,[0.1,0.1]))

print("MC means: ", np.mean(samp[:,0])*conv[0],np.mean(samp[:,1])*conv[1])
cov = np.cov(samp,rowvar=False)
cov = np.matmul(np.matmul(scale,cov),scale)
invcov = np.linalg.inv(cov)
print(invcov)
```

```python
import numpy as np
import corner
import matplotlib.pyplot as plt

# Function to read the inverse covariance matrix from a text file
def read_inverse_covariance(file_path):
    return np.loadtxt(file_path)

# Monte Carlo sampling from the multivariate normal distribution
def sample_from_inverse_covariance(inv_cov_matrix, num_samples=10000):
    # Compute the covariance matrix
    cov_matrix = np.linalg.inv(inv_cov_matrix)

    # Get the number of parameters (dimensionality)
    num_params = cov_matrix.shape[0]

    # Assume zero mean for simplicity
    mean = np.array([4.72351458e+02, -1.63784469e+01, 1.50720026e-01, 6.76102698e-01, 2.34678710e+02, 3.56751296e+01, 6.54204614e+01, 9.78331945e-0])

    # Generate samples
    samples = np.random.multivariate_normal(mean, cov_matrix, size=num_samples)
    return samples

# Main script
if __name__ == "__main__":
    # Path to the text file containing the inverse covariance matrix
    file_path = "invcov_isotensor.txt"

    # Read the inverse covariance matrix
    inv_cov_matrix = read_inverse_covariance(file_path)

    # Generate samples
    num_samples = 10000
    samples = sample_from_inverse_covariance(inv_cov_matrix, num_samples)

    # Create a corner plot
    figure = corner.corner(samples, 
                           labels=["ms","B/A","p0","M*/M","K0","J0","L0","zeta"],
                           quantiles=[0.16, 0.5, 0.84],
                           show_titles=True,
                           title_fmt=".2f",
                           title_kwargs={"fontsize": 12})

    # Show the plot
    plt.show()
```
